```python
import matplotlib as mpl
# Set font settings to embed text as editable fonts (Type 42 / TrueType)
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['font.family'] = 'sans-serif'
mpl.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']
```

```python
fontsize_dict = {"title": 7, "label": 8, "tick": 6, "legend": 6}
line_width = 1.2
line_alpha = 1

TOPLOT = {
    "Plasmid library":"Plasmid library",
    "Virus library": "Original viral library",
    '13': 'G-coated viral library',
    'BC15': 'Pseudotyped Viral library', 
}

lib2plot = dict()
for lib, libname in TOPLOT.items():
    lib2plot[libname] = libraries[lib]

cm = 1 / 2.54
fig, axes = plt.subplots(1, 2, figsize=(8 * cm, 4 * cm), dpi=150, squeeze=False)

# --- Plasmid / Virus library ---
virlib.plot_barcode_counts_and_percentage(
    lib2plot,
    label_fontsize=fontsize_dict["label"],
    tick_fontsize=fontsize_dict["tick"],
    line_alpha=line_alpha,
    line_width=line_width,
    colors=["C0", "C1", "C2", "C3", "C4"],
    ax=axes[0, 0],
)

virlib.plot_unique_label_fraction(
    lib2plot,
    stride=50,
    max_cells=1e6,
    log_scale=True,
    min_max_percent_unique_range=(0.5, 1.0),
    label_fontsize=fontsize_dict["label"],
    tick_fontsize=fontsize_dict["tick"],
    line_alpha=line_alpha,
    line_width=line_width,
    colors=["C0", "C1", "C2", "C3", "C4"],
    ax=axes[0, 1],
    show_legend=False,
)
axes[0, 1].set_xticks([1, 1e3, 1e6])
fig.tight_layout()
target_path = data_path = get_path(
    "barcode_diversity_analysis/figures/", data_root=DATA_ROOT
)
fig.savefig(target_path/"library_diversity.pdf", transparent=True, bbox_inches="tight")
```
